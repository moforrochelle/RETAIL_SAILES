```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# loading the datasets
df = pd.read_csv('retail_sales copy.csv')
df
```

```python
# printing the 1st 10 rows of the dataset
df.head(10)
```

```python
# printing the column names
df.columns
```

```python
# printing the shape of the dataset
df.shape
```

```python
# checking for missing values in the dataset
missing_values = df.isnull().sum()
missing_values
```

```python
# checking for duplicates in the dataset
duplicates_values = df.duplicated().sum()
duplicates_values
```

```python
#filling the missing values
df.ffill()
```

```python
# converting the date to date time object
df['Date'] = pd.to_datetime(df['Date'],format='mixed')
df
```

```python
#ensuring that the Quantity and UnitPrice are numeric 
df['Quantity'] = pd.to_numeric(df['Quantity'])
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'])
df

    
```

```python
# getting the Total Price using numpy vectorisation
df['Total Price'] = df['Quantity'].values * df['UnitPrice'].values
df
```

```python
#extracting the year
df['Year'] = df['Date'].dt.year
df 
```

```python
#extracting the momth
df['Month'] = df['Date'].dt.month
df
```

```python
#extracting the day of the week
df['Day of the week'] = df['Date'].dt.day_of_week
df
```

```python
# checking if the transaction was carried out on a weekday or on a weekend
df['IsWeekend'] = df['Day of the week']>=5
df
```

```python
# checking the total sales for the year
total_sales_by_year = df.groupby('Year')['Total Price'].sum()
total_sales_by_year
```

```python
# identifying the top 5 best selling products
product_sales = df.groupby('ProductCategory')['Total Price'].sum()
product_sales

```

```python
top_5_products = product_sales.nlargest(5)
top_5_products
```

```python
# identifying the month that had the highest sales( month 4 is April)
monthly_sales = df.groupby('Month')['Total Price'].sum()
highest_month_sales = monthly_sales.idxmax()
highest_amount_sales = monthly_sales.max()
print(f"The month with the highest sales was {highest_month_sales} with sales of {highest_amount_sales}")
```

```python
#identifying the most common payment method
Payment_method = df['PaymentMethod'].value_counts()
most_common_payment_method = Payment_method.index[0]
most_common_payment_method
```

```python
# calculating the order average value per store
df['Total Price'] = df['Quantity']* df['UnitPrice']
order_values = df.groupby('ProductID')['Total Price'].sum().reset_index


```

```python
#creating a line plot for monthly sales
plt.plot(df['Month'],df['Total Price'],linestyle = '-', color = 'g')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Price')
plt.show()
```

```python
#a bar chart showing the top 5 products by revenue
product_sales = df.groupby('ProductCategory')['Total Price'].sum()
product_sales
top_5_products = product_sales.nlargest(5)
top_5_products
plt.bar(top_5_products.index ,top_5_products.values, color = 'green')




```

```python
# creating a pie chart for sales distribution by payment method
sales_by_payment = df.groupby('PaymentMethod')['Total Price'].sum()
labels = sales_by_payment.index
sizes = sales_by_payment.values
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title('Sales Distribution by PaymentMethod')
plt.axis('equal')
plt.show()
```

```python
# a histogram for distribution of order values
df['Totalvalue'] = df['Quantity']* df['UnitPrice']
order_values = df.groupby('ProductID')['Totalvalue'].sum()
plt.hist(order_values , bins=28, edgecolor='pink')
plt.xlabel('order_values')
plt.ylabel('number of orders')
plt.title('Distribution of order values')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
